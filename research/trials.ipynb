```python
from fastapi import FastAPI, UploadFile
from fastapi.responses import JSONResponse
from pydantic import BaseModel
from PIL import Image
import os
```

```python
# FastAPI app
app = FastAPI()
```

```python
def process_image(image_path):
    image = Image.open(image_path)
    return "Car Damage Detected"
```

```python
@app.post('/upload-image')
async def upload_image(file: UploadFile):
    file_location = 'temp_image.jpg'
    with open(file_location, 'wb') as f:
        f.write(await file.read())
        result = process_image(file_location)
        return {'damage': result}
```

```python
from langchain_groq import ChatGroq
from langchain.schema import SystemMessage, HumanMessage
```

```python
llm = ChatGroq(
    temperature=0.7,
    model_name="llama3-70b-8192"
)
```

```python
def chat_with_groq(message: str):
    messages = [
        SystemMessage(content="You are an AI insurance assistant helping users file car accident claims."),
        HumanMessage(content=message)
    ]
    response = llm(messages)
    return response.content
```

```python
class ChatRequest(BaseModel):
    message: str
```

```python
@app.post("/chat")
async def chat(payload: ChatRequest):
    user_input = payload.message
    response = chat_with_groq(user_input)
    return {'response': response}
```
